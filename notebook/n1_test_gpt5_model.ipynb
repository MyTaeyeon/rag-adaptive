```python
import time
from openai import OpenAI

client = OpenAI()  # đọc OPENAI_API_KEY từ môi trường

MODELS = [
    "gpt-4.1-2025-04-14",
    "gpt-5.1-2025-11-13",
    "gpt-5-2025-08-07",
    "gpt-5-mini-2025-08-07",
    "gpt-5-nano-2025-08-07",
]

PROMPT = "Summarize the benefits of retrieval-augmented generation."
N_RUNS = 3

def measure_latency(model: str, prompt: str, n_runs: int = 3):
    latencies = []
    for i in range(n_runs):
        t0 = time.perf_counter()
        resp = client.chat.completions.create(
            model=model,
            messages=[{"role": "user", "content": prompt}],
            # max_tokens=200,
            # temperature=0.3,
        )
        t1 = time.perf_counter()
        latencies.append(t1 - t0)
        print(f"[{model}] run {i+1}: {latencies[-1]:.2f}s | first 60 chars: {resp.choices[0].message.content[:60]!r}")
    avg = sum(latencies) / len(latencies)
    print(f"==> {model} avg latency over {n_runs} runs: {avg:.2f}s\n")
    return latencies

all_results = {}
for m in MODELS:
    print(f"Testing {m} ...")
    all_results[m] = measure_latency(m, PROMPT, N_RUNS)

all_results
```
